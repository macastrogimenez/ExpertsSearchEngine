```python
import pandas as pd

df = pd.read_csv("drp_combined.csv")

df["orcid"] = df["source_file"].str.extract(
    r"(\d{4}-\d{4}-\d{4}-\d{3}[\dX])"
)

df.to_csv("new.csv", index=False)
```

```python
import pandas as pd
import ast

def get_rolecode(row):
    target_orcid = row["orcid"]
    
    # Convert the person string into a Python list
    try:
        people = ast.literal_eval(row["person"])
    except (ValueError, SyntaxError, TypeError):
        return None
    
    # Look through all people in this publication
    for person in people:
        if person.get("orcid") == target_orcid:
            return person.get("rolecode")
    
    return None

df["rolecode"] = df.apply(get_rolecode, axis=1)
```

```python
print(df["rolecode"].value_counts(dropna=False))
```

```python
print("Missing rolecodes:", df["rolecode"].isna().sum())
```

```python
df.to_csv("new.csv", index=False)
```
